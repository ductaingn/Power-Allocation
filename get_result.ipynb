```python
from helper import get_log_from_wandb
from process_results import plot_interface_usage, get_result_table
import matplotlib.pyplot as plt
import os
font = {'size': 20}
plt.rc('font', **font)
```

```python
PATH = '/home/nguyen/Downloads/Figures'
def get_results(run_ids:list[str], num_devices:int, save_path_prefix:str):
    run_dfs = []
    for run_id in run_ids:
        run_df, run = get_log_from_wandb(run_id, return_run=True)
        run_dfs.append(run_df)

        print(run_df["Algorithm"].iloc[0])
        plot_interface_usage(run_df, num_devices, save_path=os.path.join(save_path_prefix, f'InterfaceUsage{run_df['Algorithm'].iloc[0]}.pdf'))
    
    table = get_result_table(run_dfs, num_devices, save_path=os.path.join(save_path_prefix, 'Result.csv'))
    print(table)

    return run_dfs, table
```

# Scenario 1

## 5 dBm

```python
run_dfs, result_table = get_results(
    run_ids=[
        "42m5wqi2", # Random
        "dkkeupts", # RAQL
        "6evw1g8d", # SACPF
        "q3kna3u4", # SACPA
    ],
    num_devices=10,
    save_path_prefix=os.path.join(PATH, 'scenario1/5dbm')
)
```
